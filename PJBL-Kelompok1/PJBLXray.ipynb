```python
# ================================================================
#   KLASIFIKASI PENYAKIT PARU — CHEST X-RAY CNN
#   Dataset : COVID19 | NORMAL | PNEUMONIA | TUBERCULOSIS
#   Tools   : TensorFlow, Keras, OpenCV, Matplotlib, Sklearn
# ================================================================


# !pip install -q tensorflow opencv-python-headless matplotlib seaborn scikit-learn
# !pip install -q --upgrade gdown
# !apt-get install -y unrar -q


import os, sys, zipfile
import cv2
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, classification_report

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.utils import to_categorical

os.environ['TF_CPP_MIN_LOG_LEVEL'] = '2'

print("=" * 60)
print("  KLASIFIKASI PENYAKIT PARU — CHEST X-RAY CNN")
print("=" * 60)
print(f"  TensorFlow : {tf.__version__}")
print(f"  OpenCV     : {cv2.__version__}")
print(f"  Python     : {sys.version.split()[0]}")
print("=" * 60)


# KONFIGURASI
# ──────────────────────────────────────────────────────────────
GDRIVE_FILE_ID  = "1IBA-vq3umacLW_jExMn-wUnRxydjOJe0"  # File ID dari Google Drive
GDRIVE_FILENAME = "PJBL.rar"                             # Nama file asli (.zip / .rar)
FOLDER_INNER    = "PJBL"                                 # Nama folder di dalam arsip

IMG_SIZE    = 128
BATCH_SIZE  = 32
EPOCHS      = 15
TEST_SIZE   = 0.20
VAL_SIZE    = 0.10
RANDOM_SEED = 42

CATEGORIES  = ["COVID19", "NORMAL", "PNEUMONIA", "TUBERCULOSIS"]
NUM_CLASSES = len(CATEGORIES)

COLORS = {
    "COVID19":      "#E74C3C",
    "NORMAL":       "#2ECC71",
    "PNEUMONIA":    "#3498DB",
    "TUBERCULOSIS": "#F39C12",
}

DESKRIPSI = {
    "COVID19":      "Bercak putih bilateral (ground-glass opacity) menyebar di kedua lapang paru.",
    "NORMAL":       "Paru bersih, tidak ada bercak. Struktur bronkovaskular normal.",
    "PNEUMONIA":    "Konsolidasi/infiltrat lokal pada satu lapang paru.",
    "TUBERCULOSIS": "Lesi/kavitas pada apex (bagian atas) paru.",
}

FITUR = {
    "COVID19":      ["• Bercak bilateral", "• Ground-glass opacity", "• Menyebar merata",      "• Dominan perifer"],
    "NORMAL":       ["• Paru bersih",      "• Tidak ada infiltrat",  "• Sinus tajam",           "• Diafragma jelas"],
    "PNEUMONIA":    ["• Infiltrat lokal",  "• Konsolidasi 1 lapang", "• Bercak tidak merata",   "• Bisa air bronchogram"],
    "TUBERCULOSIS": ["• Lesi apex",        "• Bisa ada kavitas",     "• Kalsifikasi",           "• Fibrosis lama"],
}


# 1. DOWNLOAD & EKSTRAK DATASET
# ──────────────────────────────────────────────────────────────
def download_dataset():
    import gdown

    if os.path.exists(FOLDER_INNER):
        print(f"[INFO] Dataset sudah ada → skip download")
        return FOLDER_INNER

    if os.path.exists(GDRIVE_FILENAME):
        os.remove(GDRIVE_FILENAME)

    ext = os.path.splitext(GDRIVE_FILENAME)[1].lower()
    print(f"[INFO] Download {ext.upper()} dari Google Drive ...")
    gdown.download(
        f"https://drive.google.com/uc?id={GDRIVE_FILE_ID}",
        GDRIVE_FILENAME, quiet=False, fuzzy=True
    )

    print(f"[INFO] Mengekstrak {GDRIVE_FILENAME} ...")
    if ext == ".zip":
        with zipfile.ZipFile(GDRIVE_FILENAME, "r") as zf:
            zf.extractall(".")
    elif ext == ".rar":
        if os.system("which unrar > /dev/null 2>&1") != 0:
            os.system("apt-get install -y unrar -q")
        ret = os.system(f'unrar x -y "{GDRIVE_FILENAME}" ./ > /dev/null 2>&1')
        if ret != 0:
            os.system(f'unar -o . "{GDRIVE_FILENAME}"')
    else:
        raise ValueError(f"Format tidak didukung: {ext}")

    os.remove(GDRIVE_FILENAME)

    if not os.path.exists(FOLDER_INNER):
        print("\n[PERINGATAN] Folder tidak ditemukan setelah ekstrak. Folder tersedia:")
        for d in sorted(os.listdir(".")):
            if os.path.isdir(d) and not d.startswith("."):
                print(f"  📁 {d}/")
        print(f"\n→ Ganti FOLDER_INNER sesuai nama folder di atas.")
        raise FileNotFoundError(f"Folder '{FOLDER_INNER}' tidak ditemukan.")

    print(f"[OK] Dataset siap: {FOLDER_INNER}/")
    return FOLDER_INNER


DATASET_PATH = download_dataset()


# 2. MEMBACA DATASET
# ──────────────────────────────────────────────────────────────
def load_dataset(path, categories, img_size):
    images, labels, counts = [], [], {}
    print(f"\n[INFO] Membaca dataset ...")
    for idx, cat in enumerate(categories):
        folder = os.path.join(path, cat)
        if not os.path.exists(folder):
            print(f"  ✗  {cat:15s} — tidak ditemukan!")
            counts[cat] = 0
            continue
        files  = [f for f in os.listdir(folder)
                  if f.lower().endswith(('.jpg','.jpeg','.png','.bmp'))]
        loaded = 0
        for fname in files:
            img = cv2.imread(os.path.join(folder, fname))
            if img is None: continue
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img = cv2.resize(img, (img_size, img_size))
            img = img.astype(np.float32) / 255.0
            images.append(img)
            labels.append(idx)
            loaded += 1
        counts[cat] = loaded
        print(f"  ✓  {cat:15s} : {loaded:5,d} gambar")

    images = np.array(images)
    labels = np.array(labels)
    print(f"\n  Total : {len(images):,d} gambar | Shape: {images.shape}")
    return images, labels, counts


images, labels, counts = load_dataset(DATASET_PATH, CATEGORIES, IMG_SIZE)


# 3. CONTOH GAMBAR TIAP KATEGORI
# ──────────────────────────────────────────────────────────────
def tampilkan_contoh_gambar(images, labels, categories, n=3):
    fig, axes = plt.subplots(len(categories), n, figsize=(4*n, 4*len(categories)))
    fig.suptitle("Contoh Gambar X-Ray per Kategori", fontsize=16, fontweight='bold', y=1.01)
    for row, cat in enumerate(categories):
        idx_list = np.where(labels == row)[0]
        for col in range(n):
            ax = axes[row][col]
            if col < len(idx_list):
                ax.imshow(images[idx_list[col]])
                if col == 0:
                    ax.set_ylabel(cat, fontsize=12, fontweight='bold',
                                  color=COLORS[cat], rotation=0, labelpad=65, va='center')
            else:
                ax.set_visible(False)
            ax.set_xticks([]); ax.set_yticks([])
            for sp in ax.spines.values():
                sp.set_edgecolor(COLORS[cat]); sp.set_linewidth(2)
    plt.tight_layout()
    plt.savefig("01_contoh_gambar.png", dpi=120, bbox_inches='tight')
    plt.show()
    print("[OK] Disimpan: 01_contoh_gambar.png")

tampilkan_contoh_gambar(images, labels, CATEGORIES)


# 4. DISTRIBUSI DATASET
# ──────────────────────────────────────────────────────────────
def tampilkan_distribusi(counts, categories):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle("Distribusi Dataset Chest X-Ray", fontsize=14, fontweight='bold')
    vals   = [counts.get(c, 0) for c in categories]
    colors = [COLORS[c] for c in categories]

    bars = ax1.bar(categories, vals, color=colors, edgecolor='white', linewidth=1.5, width=0.55)
    ax1.set_xlabel("Kategori"); ax1.set_ylabel("Jumlah Gambar")
    ax1.set_title("Jumlah Gambar per Kategori")
    ax1.set_ylim(0, max(vals)*1.18)
    ax1.spines[['top','right']].set_visible(False)
    ax1.grid(axis='y', alpha=0.3)
    for bar, v in zip(bars, vals):
        ax1.text(bar.get_x()+bar.get_width()/2, bar.get_height()+max(vals)*0.01,
                 f"{v:,}", ha='center', fontweight='bold', fontsize=10)

    ax2.pie(vals, labels=categories, colors=colors, autopct='%1.1f%%',
            startangle=90, pctdistance=0.78,
            wedgeprops=dict(width=0.55, edgecolor='white', linewidth=2))
    ax2.set_title(f"Proporsi Dataset (Total: {sum(vals):,})")

    plt.tight_layout()
    plt.savefig("02_distribusi_dataset.png", dpi=120, bbox_inches='tight')
    plt.show()
    print("[OK] Disimpan: 02_distribusi_dataset.png")

tampilkan_distribusi(counts, CATEGORIES)


# 5. SPLIT DATASET
# ──────────────────────────────────────────────────────────────
print("\n[INFO] Split Train / Validasi / Test ...")
labels_cat = to_categorical(labels, num_classes=NUM_CLASSES)

X_tv, X_test, y_tv, y_test = train_test_split(
    images, labels_cat, test_size=TEST_SIZE,
    random_state=RANDOM_SEED, stratify=labels)

X_train, X_val, y_train, y_val = train_test_split(
    X_tv, y_tv, test_size=VAL_SIZE/(1-TEST_SIZE),
    random_state=RANDOM_SEED)

total = len(images)
print(f"  Training : {len(X_train):,d} ({len(X_train)/total*100:.0f}%)")
print(f"  Validasi : {len(X_val):,d}   ({len(X_val)/total*100:.0f}%)")
print(f"  Test     : {len(X_test):,d}   ({len(X_test)/total*100:.0f}%)")


# 6. MODEL CNN
# ──────────────────────────────────────────────────────────────
def buat_model(input_shape, num_classes):
    model = Sequential([
        Conv2D(32,  (3,3), activation='relu', padding='same', input_shape=input_shape),
        BatchNormalization(), MaxPooling2D((2,2)),
        Conv2D(64,  (3,3), activation='relu', padding='same'),
        BatchNormalization(), MaxPooling2D((2,2)),
        Conv2D(128, (3,3), activation='relu', padding='same'),
        BatchNormalization(), MaxPooling2D((2,2)),
        Flatten(),
        Dense(256, activation='relu'), Dropout(0.4),
        Dense(128, activation='relu'), Dropout(0.3),
        Dense(num_classes, activation='softmax'),
    ], name="CNN_XRay_Paru")
    model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
    return model

model = buat_model((IMG_SIZE, IMG_SIZE, 3), NUM_CLASSES)
print("\n[INFO] Arsitektur Model:")
model.summary()


# 7. TRAINING
# ──────────────────────────────────────────────────────────────
print(f"\n[INFO] Training ... (max {EPOCHS} epoch)")

history = model.fit(
    X_train, y_train,
    epochs=EPOCHS, batch_size=BATCH_SIZE,
    validation_data=(X_val, y_val),
    callbacks=[
        EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1),
        ModelCheckpoint('model_xray_best.h5', monitor='val_accuracy', save_best_only=True, verbose=1),
    ],
    verbose=1
)

model.save('model_xray.h5')
print("\n[OK] model_xray.h5       — model final")
print("[OK] model_xray_best.h5  — model terbaik")


# 8. GRAFIK TRAINING
# ──────────────────────────────────────────────────────────────
def tampilkan_grafik(history):
    acc, val_acc = history.history['accuracy'], history.history['val_accuracy']
    loss, val_loss = history.history['loss'], history.history['val_loss']
    ep = range(1, len(acc)+1)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    fig.suptitle("Grafik Training Model CNN", fontsize=14, fontweight='bold')

    ax1.plot(ep, acc,     'o-',  color='#3498DB', lw=2, ms=5, label='Train')
    ax1.plot(ep, val_acc, 's--', color='#E74C3C', lw=2, ms=5, label='Validasi')
    ax1.set_title("Accuracy"); ax1.set_xlabel("Epoch"); ax1.set_ylim(0, 1.08)
    ax1.legend(); ax1.grid(alpha=0.3); ax1.spines[['top','right']].set_visible(False)

    ax2.plot(ep, loss,     'o-',  color='#2ECC71', lw=2, ms=5, label='Train')
    ax2.plot(ep, val_loss, 's--', color='#F39C12', lw=2, ms=5, label='Validasi')
    ax2.set_title("Loss"); ax2.set_xlabel("Epoch")
    ax2.legend(); ax2.grid(alpha=0.3); ax2.spines[['top','right']].set_visible(False)

    plt.tight_layout()
    plt.savefig("03_grafik_training.png", dpi=120, bbox_inches='tight')
    plt.show()
    print("[OK] Disimpan: 03_grafik_training.png")

tampilkan_grafik(history)


# 9. EVALUASI
# ──────────────────────────────────────────────────────────────
print("\n[INFO] Evaluasi pada data test ...")
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\n  ✓ Test Accuracy : {test_acc*100:.2f}%")
print(f"  ✓ Test Loss     : {test_loss:.4f}")

y_pred = np.argmax(model.predict(X_test, verbose=0), axis=1)
y_true = np.argmax(y_test, axis=1)


# 10. CONFUSION MATRIX
# ──────────────────────────────────────────────────────────────
def tampilkan_confusion_matrix(y_true, y_pred, categories):
    cm = confusion_matrix(y_true, y_pred)
    fig, ax = plt.subplots(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=categories, yticklabels=categories,
                linewidths=0.5, linecolor='white',
                annot_kws={"size":13,"weight":"bold"}, ax=ax)
    ax.set_title("Confusion Matrix", fontsize=13, fontweight='bold', pad=15)
    ax.set_xlabel("Prediksi"); ax.set_ylabel("Aktual")
    ax.tick_params(axis='x', rotation=30)
    plt.tight_layout()
    plt.savefig("04_confusion_matrix.png", dpi=120, bbox_inches='tight')
    plt.show()
    print("[OK] Disimpan: 04_confusion_matrix.png")

tampilkan_confusion_matrix(y_true, y_pred, CATEGORIES)


# 11. CLASSIFICATION REPORT
# ──────────────────────────────────────────────────────────────
print("\n" + "="*60)
print("  CLASSIFICATION REPORT")
print("="*60)
print(classification_report(y_true, y_pred, target_names=CATEGORIES, digits=4))


# 12. PREDIKSI GAMBAR BARU
# ──────────────────────────────────────────────────────────────
def prediksi_gambar(image_path, model, categories=CATEGORIES, img_size=IMG_SIZE):
    img_raw = cv2.imread(image_path)
    if img_raw is None:
        raise FileNotFoundError(f"Gambar tidak ditemukan: '{image_path}'")

    img_rgb   = cv2.cvtColor(img_raw, cv2.COLOR_BGR2RGB)
    img_input = np.expand_dims(cv2.resize(img_rgb, (img_size, img_size)).astype(np.float32)/255.0, 0)

    probs      = model.predict(img_input, verbose=0)[0]
    pred_idx   = int(np.argmax(probs))
    pred_class = categories[pred_idx]
    confidence = float(probs[pred_idx]) * 100

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))
    fig.suptitle("Hasil Prediksi X-Ray Paru", fontsize=14, fontweight='bold')

    ax1.imshow(img_rgb)
    ax1.set_title(f"Prediksi : {pred_class}\nConfidence : {confidence:.1f}%",
                  fontsize=13, fontweight='bold', color=COLORS[pred_class])
    ax1.axis('off')

    bars = ax2.barh(categories, probs*100,
                    color=[COLORS[c] for c in categories], edgecolor='white', height=0.5)
    ax2.set_xlim(0, 115); ax2.set_xlabel("Confidence (%)")
    ax2.set_title("Probabilitas per Kelas")
    ax2.spines[['top','right']].set_visible(False)
    for bar, prob in zip(bars, probs):
        ax2.text(bar.get_width()+1.5, bar.get_y()+bar.get_height()/2,
                 f"{prob*100:.1f}%", va='center', fontsize=10,
                 fontweight='bold' if prob==probs.max() else 'normal')
    bars[pred_idx].set_edgecolor('black'); bars[pred_idx].set_linewidth(2)
    fig.text(0.5, -0.04, f"Analisis: {DESKRIPSI[pred_class]}",
             ha='center', fontsize=9, style='italic', color=COLORS[pred_class])

    plt.tight_layout()
    plt.savefig("05_hasil_prediksi.png", dpi=120, bbox_inches='tight')
    plt.show()

    print(f"\n  Prediksi   : {pred_class}")
    print(f"  Confidence : {confidence:.1f}%")
    for cat, prob in zip(categories, probs):
        bar_str = "█" * int(prob*25)
        print(f"  {cat:15s} {prob*100:5.1f}%  {bar_str}{' ◄' if prob==probs.max() else ''}")

    return pred_class, confidence


# ── Demo: prediksi 1 gambar acak dari data test ───────────────
print("\n[INFO] Demo prediksi ...")
demo_idx   = np.random.randint(0, len(X_test))
demo_label = CATEGORIES[np.argmax(y_test[demo_idx])]
demo_path  = "demo_xray.png"
cv2.imwrite(demo_path, cv2.cvtColor((X_test[demo_idx]*255).astype(np.uint8), cv2.COLOR_RGB2BGR))
print(f"  Label asli: {demo_label}")
kelas_pred, conf = prediksi_gambar(demo_path, model)
print(f"\n  {'✓ BENAR' if kelas_pred==demo_label else '✗ SALAH'}")


# 13. KARTU ANALISIS KLINIS
# ──────────────────────────────────────────────────────────────
def tampilkan_analisis_kelas(categories, colors, fitur):
    fig, axes = plt.subplots(1, len(categories), figsize=(16, 4.5))
    fig.suptitle("Karakteristik Visual X-Ray Tiap Kategori", fontsize=13, fontweight='bold')
    for ax, cat in zip(axes, categories):
        c = colors[cat]
        ax.set_facecolor(c+"18"); ax.set_xlim(0,1); ax.set_ylim(0,1); ax.axis('off')
        ax.add_patch(plt.Rectangle((0,0.80),1,0.20, color=c, alpha=0.85, transform=ax.transAxes))
        ax.text(0.5, 0.90, cat, ha='center', va='center',
                fontsize=12, fontweight='bold', color='white', transform=ax.transAxes)
        for i, item in enumerate(fitur[cat]):
            ax.text(0.08, 0.68-i*0.155, item, ha='left', va='center',
                    fontsize=9.5, color='#2C3E50', transform=ax.transAxes)
        ax.add_patch(plt.Rectangle((0,0),1,1, fill=False,
                                   edgecolor=c, linewidth=2.5, transform=ax.transAxes))
    plt.tight_layout()
    plt.savefig("06_analisis_kelas.png", dpi=120, bbox_inches='tight')
    plt.show()
    print("[OK] Disimpan: 06_analisis_kelas.png")

tampilkan_analisis_kelas(CATEGORIES, COLORS, FITUR)


# 14. RINGKASAN AKHIR
# ──────────────────────────────────────────────────────────────
ep_ran = len(history.history['accuracy'])
print("\n" + "="*60)
print("  RINGKASAN HASIL")
print("="*60)
print(f"  Total Gambar      : {len(images):,}")
print(f"  Epoch Dijalankan  : {ep_ran}/{EPOCHS}")
print(f"  Train Accuracy    : {history.history['accuracy'][-1]*100:.2f}%")
print(f"  Val Accuracy      : {history.history['val_accuracy'][-1]*100:.2f}%")
print(f"  Best Val Accuracy : {max(history.history['val_accuracy'])*100:.2f}%")
print(f"  Test Accuracy     : {test_acc*100:.2f}%")
print("="*60)
print("  Output:")
for f in ["model_xray.h5", "model_xray_best.h5",
          "01_contoh_gambar.png", "02_distribusi_dataset.png",
          "03_grafik_training.png", "04_confusion_matrix.png",
          "05_hasil_prediksi.png", "06_analisis_kelas.png"]:
    print(f"    {f}")
print("="*60)

```
